```python
from pathlib import Path
import sys

try:
    from src.config import SILVER_OUTPUT_PATH
except ImportError:
    cwd = Path().resolve()
    for p in [cwd, cwd.parent]:
        if (p / "src" / "config.py").exists():
            sys.path.insert(0, str(p))
            break
    from src.config import SILVER_OUTPUT_PATH

import pandas as pd
from IPython.display import display
```

```python
SILVER_PATH = Path(SILVER_OUTPUT_PATH)
SOURCES = {
    "veiculos": "Veículos (limpos)",
    "motoristas": "Motoristas (limpos)",
    "geocercas": "Geocercas (tabulares)",
    "viagens": "Viagens (limpas)",
    "rastreamento": "Posições GPS (limpas)",
}

for nome, desc in SOURCES.items():
    caminho = SILVER_PATH / nome
    if not caminho.exists():
        print(f"{nome}: (diretório não encontrado)")
        continue
    parquets = list(caminho.glob("*.parquet"))
    if not parquets:
        print(f"{nome}: (sem arquivos parquet)")
        continue
    df = pd.read_parquet(caminho)
    print(f"\n{'='*60}")
    print(f"{nome}  —  {desc}")
    print(f"{'='*60}")
    print(f"Linhas: {len(df):,}  |  Colunas: {len(df.columns)}")
    print()
    print("Colunas e tipos:")
    for c in df.columns:
        missing = df[c].isnull().sum()
        pct = missing / len(df) * 100
        print(f"  {c:30s}  {str(df[c].dtype):15s}  nulos={missing:>5} ({pct:5.1f}%)")
    print()
    print("Primeiras 3 linhas:")
    display(df.head(3))
```

```python
print("=== RESUMO COMPARATIVO BRONZE → SILVER ===")
print()

for nome, desc in SOURCES.items():
    bronze = Path(str(SILVER_PATH).replace("silver", "bronze")) / nome
    silver = SILVER_PATH / nome

    if not bronze.exists() or not silver.exists():
        continue

    b = pd.read_parquet(bronze)
    s = pd.read_parquet(silver)

    diff = len(b) - len(s)
    novos = sorted(set(s.columns) - set(b.columns))
    removidos = sorted(set(b.columns) - set(s.columns))

    print(f"{nome:15s}  bronze={len(b):>6}  silver={len(s):>6}  dif={diff:>+6}  "
          f"colunas novas={novos}  colunas removidas={removidos}")
```
