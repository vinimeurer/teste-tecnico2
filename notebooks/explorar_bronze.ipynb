```python
from pathlib import Path
import sys

try:
    from src.config import BRONZE_OUTPUT_PATH
except ImportError:
    cwd = Path().resolve()
    for p in [cwd, cwd.parent]:
        if (p / "src" / "config.py").exists():
            sys.path.insert(0, str(p))
            break
    from src.config import BRONZE_OUTPUT_PATH

import pandas as pd
from IPython.display import display
```

```python
BRONZE_PATH = Path(BRONZE_OUTPUT_PATH)
SOURCES = ["veiculos", "motoristas", "geocercas", "viagens", "rastreamento"]

for nome in SOURCES:
    caminho = BRONZE_PATH / nome
    if not caminho.exists():
        print(f"{nome}: (diretório não encontrado)")
        continue
    parquets = list(caminho.glob("*.parquet"))
    if not parquets:
        print(f"{nome}: (sem arquivos parquet)")
        continue
    df = pd.read_parquet(caminho)
    print(f"\n{nome}  —  {len(df)} linhas, {len(df.columns)} colunas")
    print(f"Colunas: {list(df.columns)}")
    display(df.head())
```
